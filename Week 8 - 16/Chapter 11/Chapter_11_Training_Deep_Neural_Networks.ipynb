```python
# ✅ Install library (jalankan jika belum ada)
# !pip install tensorflow matplotlib numpy pandas scikit-learn

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

print("TensorFlow version:", tf.__version__)
print("Keras version:", keras.__version__)

##############################################
# 1️⃣ Dataset - Fashion MNIST
##############################################
print("=== Loading Fashion MNIST Dataset ===")
try:
    (X_train_full, y_train_full), (X_test, y_test) = keras.datasets.fashion_mnist.load_data()
    print(f"Dataset loaded successfully!")
    print(f"Training set shape: {X_train_full.shape}")
    print(f"Test set shape: {X_test.shape}")

    # Normalize pixel values to 0-1 range
    X_train_full = X_train_full / 255.0
    X_test = X_test / 255.0

    # Split training data into train and validation sets
    X_valid, X_train = X_train_full[:5000], X_train_full[5000:]
    y_valid, y_train = y_train_full[:5000], y_train_full[5000:]

    print(f"Train set: {X_train.shape}, Validation set: {X_valid.shape}")

    # Display class names
    class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
                   "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]

    # Visualize some samples
    plt.figure(figsize=(12, 8))
    for i in range(12):
        plt.subplot(3, 4, i + 1)
        plt.imshow(X_train[i], cmap='gray')
        plt.title(f"{class_names[y_train[i]]}")
        plt.axis('off')
    plt.suptitle("Fashion MNIST Samples")
    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"Error loading dataset: {e}")

##############################################
# 2️⃣ Deep Neural Network with Batch Normalization
##############################################
print("=== Building Deep Neural Network ===")
try:
    # Clear any previous models
    keras.backend.clear_session()

    model = keras.models.Sequential([
        keras.layers.Flatten(input_shape=[28, 28]),
        keras.layers.BatchNormalization(),
        keras.layers.Dense(300, activation="elu", kernel_initializer="he_normal"),
        keras.layers.BatchNormalization(),
        keras.layers.Dense(100, activation="elu", kernel_initializer="he_normal"),
        keras.layers.BatchNormalization(),
        keras.layers.Dense(10, activation="softmax")
    ])

    # Compile the model
    model.compile(loss="sparse_categorical_crossentropy",
                  optimizer=keras.optimizers.Adam(learning_rate=0.001),
                  metrics=["accuracy"])

    # Display model architecture
    print("Model Architecture:")
    model.summary()

    # Train the model
    print("Training the model...")
    history = model.fit(X_train, y_train,
                       epochs=15,  # Reduced epochs for faster training
                       batch_size=128,
                       validation_data=(X_valid, y_valid),
                       verbose=1)

    print("Training completed!")

except Exception as e:
    print(f"Error in model training: {e}")

##############################################
# 3️⃣ Learning Curves Visualization
##############################################
print("=== Visualizing Learning Curves ===")
try:
    # Plot training history
    history_df = pd.DataFrame(history.history)

    plt.figure(figsize=(15, 5))

    # Plot loss
    plt.subplot(1, 3, 1)
    plt.plot(history_df['loss'], label='Training Loss', linewidth=2)
    plt.plot(history_df['val_loss'], label='Validation Loss', linewidth=2)
    plt.title('Model Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True, alpha=0.3)

    # Plot accuracy
    plt.subplot(1, 3, 2)
    plt.plot(history_df['accuracy'], label='Training Accuracy', linewidth=2)
    plt.plot(history_df['val_accuracy'], label='Validation Accuracy', linewidth=2)
    plt.title('Model Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True, alpha=0.3)

    # Plot both metrics together
    plt.subplot(1, 3, 3)
    plt.plot(history_df.index, history_df['loss'], 'b-', label='Train Loss')
    plt.plot(history_df.index, history_df['val_loss'], 'r-', label='Val Loss')
    plt.plot(history_df.index, history_df['accuracy'], 'b--', label='Train Acc')
    plt.plot(history_df.index, history_df['val_accuracy'], 'r--', label='Val Acc')
    plt.title('Training Metrics Overview')
    plt.xlabel('Epoch')
    plt.ylabel('Value')
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    # Print final metrics
    final_train_acc = history_df['accuracy'].iloc[-1]
    final_val_acc = history_df['val_accuracy'].iloc[-1]
    print(f"Final Training Accuracy: {final_train_acc:.4f}")
    print(f"Final Validation Accuracy: {final_val_acc:.4f}")

except Exception as e:
    print(f"Error in visualization: {e}")

##############################################
# 4️⃣ Gradient Clipping Example
##############################################
print("=== Gradient Clipping Example ===")
try:
    # Create a new model with gradient clipping
    model_clipped = keras.models.Sequential([
        keras.layers.Flatten(input_shape=[28, 28]),
        keras.layers.BatchNormalization(),
        keras.layers.Dense(300, activation="elu", kernel_initializer="he_normal"),
        keras.layers.BatchNormalization(),
        keras.layers.Dense(100, activation="elu", kernel_initializer="he_normal"),
        keras.layers.BatchNormalization(),
        keras.layers.Dense(10, activation="softmax")
    ])

    # Compile with gradient clipping
    model_clipped.compile(loss="sparse_categorical_crossentropy",
                         optimizer=keras.optimizers.SGD(learning_rate=0.01, clipnorm=1.0),
                         metrics=["accuracy"])

    print("Model with gradient clipping compiled successfully!")
    print("Optimizer: SGD with clipnorm=1.0")

except Exception as e:
    print(f"Error in gradient clipping setup: {e}")

##############################################
# 5️⃣ Transfer Learning Example (Conceptual)
##############################################
print("=== Transfer Learning Concept ===")
try:
    # Create a base model (simulate pre-trained model)
    base_model = keras.models.Sequential([
        keras.layers.Flatten(input_shape=[28, 28]),
        keras.layers.Dense(128, activation="relu", name="feature_extractor"),
        keras.layers.Dense(64, activation="relu", name="intermediate_features")
    ])

    # Create a new model using base model layers (transfer learning simulation)
    transfer_model = keras.models.Sequential([
        base_model,
        keras.layers.Dense(10, activation="softmax", name="new_classifier")
    ])

    print("Transfer learning model created:")
    transfer_model.summary()

    # Demonstrate freezing layers
    for layer in base_model.layers:
        layer.trainable = False

    print("Base model layers frozen for transfer learning!")

except Exception as e:
    print(f"Error in transfer learning example: {e}")

##############################################
# 6️⃣ Optimizer Comparison
##############################################
print("=== Optimizer Examples ===")
try:
    # Different optimizers (fixed deprecated 'lr' parameter)
    optimizers = {
        'SGD': keras.optimizers.SGD(learning_rate=0.01),
        'SGD_Momentum': keras.optimizers.SGD(learning_rate=0.01, momentum=0.9),
        'SGD_Nesterov': keras.optimizers.SGD(learning_rate=0.01, momentum=0.9, nesterov=True),
        'Adam': keras.optimizers.Adam(learning_rate=0.001),
        'RMSprop': keras.optimizers.RMSprop(learning_rate=0.001),
        'Nadam': keras.optimizers.Nadam(learning_rate=0.001),
        'AdaGrad': keras.optimizers.Adagrad(learning_rate=0.01),
        'Adamax': keras.optimizers.Adamax(learning_rate=0.002)
    }

    print("Available optimizers:")
    for name, optimizer in optimizers.items():
        print(f"✓ {name}: {optimizer.__class__.__name__}")

    # Demonstrate optimizer comparison with small training
    print("\nComparing optimizer performance (3 epochs each):")
    optimizer_results = {}

    for name, optimizer in list(optimizers.items())[:3]:  # Test first 3 optimizers
        print(f"\nTesting {name}...")

        # Create fresh model
        test_model = keras.models.Sequential([
            keras.layers.Flatten(input_shape=[28, 28]),
            keras.layers.Dense(128, activation="relu"),
            keras.layers.Dense(10, activation="softmax")
        ])

        test_model.compile(loss="sparse_categorical_crossentropy",
                          optimizer=optimizer,
                          metrics=["accuracy"])

        # Quick training
        hist = test_model.fit(X_train[:1000], y_train[:1000],
                             epochs=3,
                             validation_data=(X_valid[:500], y_valid[:500]),
                             verbose=0)

        final_acc = hist.history['val_accuracy'][-1]
        optimizer_results[name] = final_acc
        print(f"{name} final validation accuracy: {final_acc:.4f}")

    # Plot comparison
    plt.figure(figsize=(10, 6))
    names = list(optimizer_results.keys())
    accuracies = list(optimizer_results.values())

    plt.bar(names, accuracies, color=['skyblue', 'lightcoral', 'lightgreen'])
    plt.title('Optimizer Comparison (3 epochs)')
    plt.ylabel('Validation Accuracy')
    plt.ylim(0, 1)
    for i, acc in enumerate(accuracies):
        plt.text(i, acc + 0.01, f'{acc:.3f}', ha='center')
    plt.show()

except Exception as e:
    print(f"Error in optimizer comparison: {e}")

##############################################
# 7️⃣ Learning Rate Scheduling
##############################################
print("=== Learning Rate Scheduling ===")
try:
    # Define learning rate scheduler functions
    def exponential_decay_fn(epoch, lr):
        return lr * 0.95

    def step_decay_fn(epoch, lr):
        if epoch % 5 == 0 and epoch > 0:
            return lr * 0.5
        return lr

    def cosine_decay_fn(epoch, lr):
        import math
        return lr * (1 + math.cos(math.pi * epoch / 10)) / 2

    # Create scheduler callbacks
    exp_scheduler = keras.callbacks.LearningRateScheduler(exponential_decay_fn)
    step_scheduler = keras.callbacks.LearningRateScheduler(step_decay_fn)
    cosine_scheduler = keras.callbacks.LearningRateScheduler(cosine_decay_fn)

    # Also create ReduceLROnPlateau scheduler
    reduce_lr = keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss', factor=0.5, patience=2, min_lr=1e-6
    )

    print("Learning rate schedulers created:")
    print("✓ Exponential decay")
    print("✓ Step decay")
    print("✓ Cosine decay")
    print("✓ ReduceLROnPlateau")

    # Demonstrate with exponential decay
    print("\nTraining with exponential decay scheduler...")

    # Create a fresh model
    scheduler_model = keras.models.Sequential([
        keras.layers.Flatten(input_shape=[28, 28]),
        keras.layers.Dense(128, activation="relu"),
        keras.layers.Dense(10, activation="softmax")
    ])

    scheduler_model.compile(loss="sparse_categorical_crossentropy",
                           optimizer=keras.optimizers.Adam(learning_rate=0.01),
                           metrics=["accuracy"])

    # Train with scheduler
    scheduler_history = scheduler_model.fit(
        X_train[:2000], y_train[:2000],
        epochs=10,
        validation_data=(X_valid[:500], y_valid[:500]),
        callbacks=[exp_scheduler],
        verbose=1
    )

    # Plot learning rate changes
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    learning_rates = [0.01 * (0.95 ** epoch) for epoch in range(10)]
    plt.plot(learning_rates, 'o-', linewidth=2, markersize=8)
    plt.title('Learning Rate Schedule (Exponential Decay)')
    plt.xlabel('Epoch')
    plt.ylabel('Learning Rate')
    plt.grid(True, alpha=0.3)

    plt.subplot(1, 2, 2)
    plt.plot(scheduler_history.history['accuracy'], label='Training')
    plt.plot(scheduler_history.history['val_accuracy'], label='Validation')
    plt.title('Training with LR Scheduling')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

except Exception as e:
    print(f"Error in learning rate scheduling: {e}")

##############################################
# 8️⃣ Model Saving and Loading
##############################################
print("=== Model Saving and Loading ===")
try:
    # Save the trained model
    model_filename = "my_deep_model.h5"
    model.save(model_filename)
    print(f"✓ Model saved as '{model_filename}'")

    # Save in SavedModel format (recommended)
    model.save("my_deep_model_savedformat")
    print("✓ Model saved in SavedModel format")

    # Load the model
    loaded_model = keras.models.load_model(model_filename)
    print("✓ Model loaded successfully")

    # Verify loaded model works
    test_prediction = loaded_model.predict(X_test[:5])
    print(f"✓ Loaded model prediction shape: {test_prediction.shape}")

    # Compare original vs loaded model predictions
    original_pred = model.predict(X_test[:5])
    loaded_pred = loaded_model.predict(X_test[:5])

    difference = np.mean(np.abs(original_pred - loaded_pred))
    print(f"✓ Prediction difference (should be ~0): {difference:.2e}")

    # Save model weights only
    model.save_weights("my_model_weights.h5")
    print("✓ Model weights saved")

    # Model configuration
    model_config = model.to_json()
    print("✓ Model configuration extracted")

    print("\nModel file sizes:")
    import os
    if os.path.exists(model_filename):
        size_mb = os.path.getsize(model_filename) / (1024*1024)
        print(f"Complete model: {size_mb:.2f} MB")

except Exception as e:
    print(f"Error in model saving/loading: {e}")

##############################################
# 9️⃣ Model Evaluation and Testing
##############################################
print("=== Final Model Evaluation ===")
try:
    # Evaluate on test set
    test_loss, test_accuracy = model.evaluate(X_test, y_test, verbose=0)
    print(f"Test Loss: {test_loss:.4f}")
    print(f"Test Accuracy: {test_accuracy:.4f}")

    # Make predictions
    predictions = model.predict(X_test[:10])
    predicted_classes = np.argmax(predictions, axis=1)

    # Show some predictions
    plt.figure(figsize=(15, 6))
    for i in range(10):
        plt.subplot(2, 5, i + 1)
        plt.imshow(X_test[i], cmap='gray')
        plt.title(f'True: {class_names[y_test[i]]}\nPred: {class_names[predicted_classes[i]]}')
        plt.axis('off')

    plt.suptitle('Model Predictions on Test Set')
    plt.tight_layout()
    plt.show()

    # Confusion matrix
    from sklearn.metrics import classification_report, confusion_matrix

    y_pred_all = np.argmax(model.predict(X_test), axis=1)

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred_all, target_names=class_names))

except Exception as e:
    print(f"Error in final evaluation: {e}")

print("\n=== Deep Learning Pipeline Completed Successfully! ===")
```

# 📚 Chapter 11 - Training Deep Neural Networks

---

## 🔸 1. Tantangan Training Neural Networks
Melatih neural network yang dalam tidak mudah karena:

| Masalah                 | Dampak                                      |
| ----------------------- | ------------------------------------------- |
| **Vanishing Gradients** | Gradien terlalu kecil → layer awal sulit belajar |
| **Exploding Gradients** | Gradien terlalu besar → training tidak stabil |

---

## 🔸 2. Strategi Inisialisasi
Pemilihan inisialisasi bobot sangat penting untuk stabilitas training.

| Metode               | Digunakan Saat                               |
| -------------------- | -------------------------------------------- |
| **Glorot/Xavier**    | Untuk aktivasi `tanh` atau `sigmoid`        |
| **He Initialization** | Untuk aktivasi `ReLU`, `ELU`, dll           |

---

## 🔸 3. Fungsi Aktivasi
| Activation | Keterangan                              |
| ---------- | --------------------------------------- |
| **ReLU**   | Cepat & stabil, standar hidden layer    |
| **ELU**    | Lebih baik untuk deep network           |
| **Leaky ReLU** | Mengurangi masalah dead neurons     |
| **SELU**   | Untuk Self-Normalizing Network (SNN)    |

---

## 🔸 4. Batch Normalization (BN)
**Batch Normalization (BN)** menormalkan output layer sebelum aktivasi → mempercepat training dan stabilisasi.

- Mengurangi **vanishing/exploding gradients**
- Efek samping positif → **regularisasi** → membantu mencegah overfitting

---

## 🔸 5. Gradient Clipping
Untuk mencegah exploding gradients → gradien dibatasi:


optimizer = keras.optimizers.SGD(clipnorm=1.0)

---

## 🔸 6. Optimizer
| Optimizer  | Keterangan                          |
| ---------- | ----------------------------------- |
| `SGD`      | Sederhana → baseline                |
| `Momentum` | Menambahkan percepatan              |
| `Nesterov` | Lebih akurat dibanding Momentum     |
| `Adam`     | **Pilihan standar** → cepat, stabil |
| `RMSprop`  | Cocok untuk sequence, RNN           |

## 🔸 7. Learning Rate Scheduling
Learning rate berubah selama training agar hasil lebih baik:

| Strategi              | Keterangan                         |
| --------------------- | ---------------------------------- |
| **Exponential Decay** | Learning rate turun terus menerus  |
| **Step Decay**        | Learning rate turun bertahap       |
| **Performance-based** | Turun saat validation loss stagnan |

## 🔸 8. Transfer Learning
Menggunakan model neural network yang telah dilatih → lanjutkan training untuk tugas berbeda → mempercepat training dan sering menghasilkan performa lebih baik.
