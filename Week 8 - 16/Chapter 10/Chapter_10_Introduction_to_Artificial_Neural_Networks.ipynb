````python
# ✅ Install library (opsional jika belum ada)
# !pip install tensorflow matplotlib numpy

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

##############################################
# 1️⃣ Dataset Fashion MNIST (Klasifikasi)
##############################################
fashion_mnist = keras.datasets.fashion_mnist
(X_train_full, y_train_full), (X_test, y_test) = fashion_mnist.load_data()

# Split train/validation + Normalisasi
X_valid, X_train = X_train_full[:5000] / 255.0, X_train_full[5000:] / 255.0
y_valid, y_train = y_train_full[:5000], y_train_full[5000:]

class_names = ["T-shirt/top", "Trouser", "Pullover", "Dress", "Coat",
               "Sandal", "Shirt", "Sneaker", "Bag", "Ankle boot"]

##############################################
# 2️⃣ Membangun Model Sequential API
##############################################
model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28, 28]),
    keras.layers.Dense(300, activation="relu"),
    keras.layers.Dense(100, activation="relu"),
    keras.layers.Dense(10, activation="softmax")
])

model.compile(loss="sparse_categorical_crossentropy",
              optimizer="sgd",
              metrics=["accuracy"])

##############################################
# 3️⃣ Training Model
##############################################
history = model.fit(X_train, y_train, epochs=30,
                    validation_data=(X_valid, y_valid))

# Plot Learning Curves
pd.DataFrame(history.history).plot(figsize=(8, 5))
plt.grid(True)
plt.gca().set_ylim(0, 1)
plt.show()

##############################################
# 4️⃣ Evaluasi Model
##############################################
model.evaluate(X_test, y_test)

##############################################
# 5️⃣ Prediksi
##############################################
X_new = X_test[:3]
y_proba = model.predict(X_new)
y_pred = np.argmax(y_proba, axis=-1)
print("Predicted classes:", y_pred)
print("Actual classes:   ", y_test[:3])
print("Predicted names:  ", np.array(class_names)[y_pred])

##############################################
# 6️⃣ Visualisasi Prediksi
##############################################
plt.figure(figsize=(10,3))
for idx, image in enumerate(X_new):
    plt.subplot(1,3,idx+1)
    plt.imshow(image, cmap="binary")
    plt.title(class_names[y_pred[idx]])
    plt.axis("off")
plt.show()

##############################################
# 7️⃣ Regression MLP - California Housing Dataset
##############################################
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()
X_train_full, X_test, y_train_full, y_test = train_test_split(housing.data, housing.target)
X_train, X_valid, y_train, y_valid = train_test_split(X_train_full, y_train_full)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_valid = scaler.transform(X_valid)
X_test = scaler.transform(X_test)

reg_model = keras.models.Sequential([
    keras.layers.Dense(30, activation="relu", input_shape=X_train.shape[1:]),
    keras.layers.Dense(1)
])

reg_model.compile(loss="mean_squared_error", optimizer="sgd")
reg_history = reg_model.fit(X_train, y_train, epochs=20, validation_data=(X_valid, y_valid))

reg_model.evaluate(X_test, y_test)

##############################################
# 8️⃣ Prediksi Housing
##############################################
X_new_reg = X_test[:3]
y_pred_reg = reg_model.predict(X_new_reg)
print("Prediksi harga:", y_pred_reg.flatten())
print("Harga asli:    ", y_test[:3])
````

# 📚 Chapter 10 - Introduction to Artificial Neural Networks (ANNs)

---

## 🔸 1. Apa Itu Artificial Neural Networks (ANN)?
Artificial Neural Networks (ANNs) adalah model komputasi terinspirasi dari cara kerja otak → terdiri dari neuron buatan yang saling terhubung.

---

## 🔸 2. Struktur ANN
| Komponen         | Keterangan                                       |
| ---------------- | ------------------------------------------------ |
| **Neuron (Perceptron)** | Unit dasar → menghitung weighted sum + activation |
| **Layer Input**  | Menerima input fitur                             |
| **Hidden Layers**| Memproses input → tempat ANN belajar pola        |
| **Output Layer** | Menghasilkan prediksi akhir                     |

---

## 🔸 3. Perceptron
Model ANN paling dasar → linear classifier:

\[
\hat{y} = f(\mathbf{w} \cdot \mathbf{x} + b)
\]
- **f** = fungsi aktivasi (misalnya: step, sigmoid, relu)

---

## 🔸 4. Multilayer Perceptron (MLP)
ANN dengan **satu atau lebih hidden layers** → mampu mempelajari hubungan **non-linear**.

**Fungsi Aktivasi Umum:**
| Activation | Ciri                                |
| ---------- | ---------------------------------- |
| `relu`     | Cepat, cocok untuk hidden layers   |
| `softmax`  | Untuk output multi-class classification |

---

## 🔸 5. Training ANN
ANN belajar dengan cara **menyesuaikan bobot** untuk meminimalkan fungsi biaya (**loss function**).

**Algoritma:**
1. **Forward Propagation** → hitung output prediksi
2. **Loss Function** → hitung error (misalnya: MSE, Cross-Entropy)
3. **Backpropagation** → hitung gradient error terhadap bobot
4. **Gradient Descent** → update bobot agar error ↓

---

## 🔸 6. Loss Function
| Loss                         | Kapan digunakan                          |
| ---------------------------- | ---------------------------------------- |
| `mean_squared_error (MSE)`    | Untuk **regresi**                       |
| `sparse_categorical_crossentropy` | Untuk **klasifikasi multi-kelas integer** |

---

## 🔸 7. Optimizers
| Optimizer  | Ciri                        |
| ---------- | -------------------------- |
| `SGD`      | Stochastic Gradient Descent (sederhana) |
| `Adam`     | Lebih cepat konvergen → sering dipakai |

---

## 🔸 8. Dataset Contoh
- ✅ **Fashion MNIST** → klasifikasi gambar pakaian → 10 kelas
- ✅ **California Housing** → regresi → memprediksi harga rumah
