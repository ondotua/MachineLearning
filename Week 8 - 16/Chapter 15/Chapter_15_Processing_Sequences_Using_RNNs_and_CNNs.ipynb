```python
# ✅ Install library jika diperlukan
# !pip install tensorflow tensorflow-datasets numpy matplotlib

import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt

##############################################
# 1️⃣ Simple RNN Example
##############################################
np.random.seed(42)
tf.random.set_seed(42)

series = np.sin(0.1 * np.arange(10000)) + np.random.normal(scale=0.1, size=10000)

def windowed_dataset(series, window_size, batch_size=32, shuffle_buffer=1000):
    ds = tf.data.Dataset.from_tensor_slices(series)
    ds = ds.window(window_size + 1, shift=1, drop_remainder=True)
    ds = ds.flat_map(lambda w: w.batch(window_size + 1))
    ds = ds.shuffle(shuffle_buffer)
    ds = ds.map(lambda w: (w[:-1], w[-1]))
    return ds.batch(batch_size).prefetch(1)

window_size = 20
dataset = windowed_dataset(series, window_size)

model = keras.models.Sequential([
    keras.layers.SimpleRNN(40, return_sequences=True, input_shape=[None, 1]),
    keras.layers.SimpleRNN(40),
    keras.layers.Dense(1)
])

model.compile(loss="mse", optimizer="adam")
model.fit(dataset, epochs=5)

##############################################
# 2️⃣ Forecasting with RNN
##############################################
def model_forecast(model, series, window_size):
    ds = tf.data.Dataset.from_tensor_slices(series)
    ds = ds.window(window_size, shift=1, drop_remainder=True)
    ds = ds.flat_map(lambda w: w.batch(window_size))
    ds = ds.batch(32).prefetch(1)
    return model.predict(ds)

forecast = model_forecast(model, series[..., np.newaxis], window_size)
forecast = forecast[:-1, 0]
plt.plot(series[window_size:], label="Target")
plt.plot(forecast, label="Forecast")
plt.legend()
plt.grid(True)
plt.show()

##############################################
# 3️⃣ Using LSTM
##############################################
model = keras.models.Sequential([
    keras.layers.LSTM(20, return_sequences=True, input_shape=[None, 1]),
    keras.layers.LSTM(20),
    keras.layers.Dense(1)
])

model.compile(loss="mse", optimizer="adam")
model.fit(dataset, epochs=5)

##############################################
# 4️⃣ Bidirectional RNN
##############################################
model = keras.models.Sequential([
    keras.layers.Bidirectional(keras.layers.LSTM(20, return_sequences=True), input_shape=[None, 1]),
    keras.layers.Bidirectional(keras.layers.LSTM(20)),
    keras.layers.Dense(1)
])

model.compile(loss="mse", optimizer="adam")
model.fit(dataset, epochs=5)

##############################################
# 5️⃣ Sequence Classification (Example: IMDB)
##############################################
(X_train, y_train), (X_test, y_test) = keras.datasets.imdb.load_data(num_words=10000)
X_train = keras.preprocessing.sequence.pad_sequences(X_train, maxlen=500)
X_test = keras.preprocessing.sequence.pad_sequences(X_test, maxlen=500)

model = keras.models.Sequential([
    keras.layers.Embedding(input_dim=10000, output_dim=32, input_length=500),
    keras.layers.Bidirectional(keras.layers.LSTM(32)),
    keras.layers.Dense(1, activation="sigmoid")
])

model.compile(loss="binary_crossentropy", optimizer="adam", metrics=["accuracy"])
history = model.fit(X_train, y_train, epochs=3, validation_split=0.2)

##############################################
# 6️⃣ Sequence-to-Sequence Example (Toy)
##############################################
encoder_inputs = keras.layers.Input(shape=[None, 1])
encoder = keras.layers.LSTM(32, return_state=True)
encoder_outputs, state_h, state_c = encoder(encoder_inputs)
encoder_state = [state_h, state_c]

decoder_inputs = keras.layers.Input(shape=[None, 1])
decoder_lstm = keras.layers.LSTM(32, return_sequences=True, return_state=True)
decoder_outputs, _, _ = decoder_lstm(decoder_inputs, initial_state=encoder_state)
decoder_dense = keras.layers.Dense(1)
decoder_outputs = decoder_dense(decoder_outputs)

seq2seq_model = keras.models.Model(inputs=[encoder_inputs, decoder_inputs], outputs=[decoder_outputs])
seq2seq_model.summary()
```

# 📚 Chapter 15 - Processing Sequences Using RNNs and CNNs

---

## 🔸 1. Apa Itu RNN?
**Recurrent Neural Networks (RNNs)** dirancang untuk memproses **data berurutan (sequence data)**, contohnya:
- Time Series (misalnya suhu harian, harga saham)
- Teks (natural language)
- Audio (waveform)

---

## 🔸 2. Cara Kerja RNN
- Pada setiap langkah waktu `t`, input **x(t)** diproses bersama dengan **output sebelumnya y(t-1)** → memberi efek **memori jangka pendek**.
- **Backpropagation Through Time (BPTT)** digunakan untuk melatih model.
- ⚠ **Masalah:** rentan terhadap **vanishing/exploding gradients** → sulit untuk sequence panjang.

---

## 🔸 3. Variasi Input & Output
| Tipe                 | Contoh Penggunaan         |
| -------------------- | ------------------------ |
| Sequence → Sequence  | Prediksi harga saham     |
| Sequence → Vector    | Sentiment analysis       |
| Vector → Sequence    | Image caption generation |
| Encoder → Decoder    | Machine Translation      |

---

## 🔸 4. Memori Jangka Panjang
✅ Solusi untuk **short-term memory problem**:
- **LSTM (Long Short-Term Memory)**
- **GRU (Gated Recurrent Unit)**
→ Lebih stabil untuk sequence panjang.

---

## 🔸 5. Arsitektur RNN Modern
| Layer                | Fungsi                                |
| -------------------- | ------------------------------------- |
| SimpleRNN            | RNN dasar → cocok untuk demo          |
| LSTM                 | Lebih baik, tahan long-term dependen. |
| GRU                  | Lebih efisien, performa hampir setara LSTM |

---

## 🔸 6. Handling Long Sequences
| Teknik              | Fungsi                            |
| ------------------- | --------------------------------- |
| Gradient Clipping   | Mencegah exploding gradients     |
| Recurrent Dropout   | Regularisasi                     |
| Layer Normalization | Stabilkan training                |

---

## 🔸 7. 1D Convolution for Sequences
✅ CNN bisa digunakan untuk sequence → **1D Convolution**:
- Memperpendek input sequence → deteksi **pola jangka panjang lebih efisien**
- Basis dari model seperti **WaveNet** → sangat cocok untuk audio/text sequence panjang.

---

## 🔸 8. WaveNet (DeepMind)
- Stack Conv1D dengan **dilated convolutions** → belajar pola pendek & panjang.
- Digunakan oleh DeepMind → text-to-speech, music generation.

---

## ✅ Kesimpulan
- ✅ Gunakan RNN (LSTM/GRU) untuk prediksi data berurutan.
- ✅ Gunakan **1D CNN atau WaveNet** → sequence panjang atau audio.
- ✅ Untuk NLP modern → lanjutkan ke **Transformer** (Chapter 16).
